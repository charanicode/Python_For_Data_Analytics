```python
Name_student="Charani"
Age=21
is_a_graduate=True
Salary=0.0

print(Name_student)
print(Age)
print(is_a_graduate)
print(Salary)

#to know the type of variable,use the type()
print(type(Name_student))
print(type(Age))
print(type(is_a_graduate))
print(type(Salary))
```

```python
#arithmetic operators:+,-,*,/,//,%%
a=5
b=2
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
```

```python
#Assignment Operators:+=,-=,*=,/=,//=,%=
x=5
x+=2
print(x)
x-=3
print(x)
x*=4
print(x)
x/=2
print(x)
x//=4
print(x)
x%=2
print(x)
```

```python
#Comparision Operators
x=20
y=10
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x<=y)
print(x>=y)
print(2<5>3)
print(3<4>4)
```
