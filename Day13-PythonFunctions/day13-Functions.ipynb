```python
def displaydetails(name,age):
    print("Hello ",name,'Your Age is ',age)
displaydetails("Raghu",22)
displaydetails(22,"Raghu")
displaydetails(name="Raghu",age=22)
displaydetails(age=22,name="Raghu")
```

```python
def shownames(*names):
    for name in names:
        print("Good Morning ",name)
    print("Total Names in function are ",len(names),"And type are ",type(names))
shownames("Charani","Charan","Baron")
```

```python
#recursion:
def fact(n):
    if(n==0 | n==1):
        return 1
    return n*fact(n-1)
fact(5)
```

```python
def printnumbers(n):
    if n<=5:
        print(n)
        printnumbers(n+1)
printnumbers(1)
```
