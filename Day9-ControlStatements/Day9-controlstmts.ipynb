```python
#check if a number is positive or negative
num=int(input("enter a number: "))
if(num>0):
    print(num,"is a positive number")
elif(num==0):
    print(num,"is a zero")
else:
    print(num,"is a negative number")

```

```python
#LeapYear
year=int(input("enter a year "))
if (year %4==0 and year %100 !=0) or year%400==0:
    print("Leap Year")
else:
    print("Not a leap year")
```

```python
#3.check a number is divisble by 3 and 5
num=int(input("enter a number: "))
if num % 3==0 and num % 5==0:
    print(num,"is divisble by both 3 and 5")
elif num%3==0:
    print(num,"is divisble by 3")
elif num%5==0:
    print(num,"is divisble by 5")
else:
    print(num,"is not divisble by both 3 and 5")
```

```python
#check if a number is a multiple of 2 and 3
num=int(input("enter a number: "))
if num%2==0 and num %3==0:
    print("It is a multiple of 2 and 3")
elif num%2==0:
    print("It is a multiple of 2")
elif num%3==0:
    print("It is a multiple of 3")
else:
    print("it is not a multiple of 2 and 3")
```

```python
#Student Grade
marks=int(input("enter the marks:"))
if marks>90:
    print("congratulations,you scored Grade A")
elif marks>80 and marks<=90:
    print("You have scored grade B")
elif marks>60 and marks<=80:
    print("You have scored grade C")
else:
    print("You have scored grade F")
```
